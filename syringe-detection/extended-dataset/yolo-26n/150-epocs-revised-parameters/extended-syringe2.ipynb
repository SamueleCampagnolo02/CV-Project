```python
import yaml

data_yaml_path = "/kaggle/working/custom_syringe_yolo/data.yaml"
with open(data_yaml_path, "r") as f:
    cfg = yaml.safe_load(f)
cfg["path"] = "/kaggle/working/custom_syringe_yolo"
with open(data_yaml_path, "w") as f:
    yaml.dump(cfg, f)
print(cfg)
```

```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")

results = model.train(
    data="/kaggle/working/custom_syringe_yolo/data.yaml",
    epochs=150,
    imgsz=640,
    batch=16,
    device=0,
    mixup=0.1,
    copy_paste=0.2,
    degrees=15,
    scale=0.7,
    hsv_v=0.3,
    project="/kaggle/working/runs_syringe_scratch",
    name="yolo26n_syringe_scratch_150epochs",
)
```
